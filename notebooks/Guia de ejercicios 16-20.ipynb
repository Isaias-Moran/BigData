```python
#Ejercicio 17: Análisis de ventas con pandas
#• Cree un DataFrame de ventas con al menos 15 registros.
#• Incluya fecha, vendedor, producto, categoría, cantidad y precio unitario.
#• Calcule una nueva columna TotalVenta.
#• Calcule el total vendido por vendedor.
#• Identifique el producto más vendido.
#• Identifique la categoría con mayor ingreso.
#• Ordene las ventas de mayor a menor.


import pandas as pd


datos_ventas = {
    "fecha": ["2026-05-01", "2026-05-01", "2026-05-02", "2026-05-02", "2026-05-03", "2026-05-03", "2026-05-04", "2026-05-04", "2026-05-05", "2026-05-05", "2026-05-06", "2026-05-06", "2026-05-07", "2026-05-07", "2026-05-08"],
    "vendedor": ["Ana", "Carlos", "Ana", "Luis", "Carlos", "Luis", "Ana", "Carlos", "Ana", "Luis", "Carlos", "Luis", "Ana", "Carlos", "Luis"],
    "producto": ["Mouse", "Teclado", "Monitor", "Mouse", "Laptop", "Teclado", "Monitor", "Mouse", "Teclado", "Laptop", "Mouse", "Monitor", "Teclado", "Laptop", "Mouse"],
    "categoría": ["Accesorios", "Accesorios", "Pantallas", "Accesorios", "Cómputo", "Accesorios", "Pantallas", "Accesorios", "Accesorios", "Cómputo", "Accesorios", "Pantallas", "Accesorios", "Cómputo", "Accesorios"],
    "cantidad": [2, 1, 1, 5, 1, 2, 2, 3, 1, 1, 4, 1, 2, 1, 3],
    "precio_unitario": [15, 25, 150, 15, 800, 25, 150, 15, 25, 800, 15, 150, 25, 800, 15]
}

df_ventas = pd.DataFrame(datos_ventas)

# 2. Calcule una nueva columna TotalVenta
df_ventas["TotalVenta"] = df_ventas["cantidad"] * df_ventas["precio_unitario"]

# 3. Calcule el total vendido por vendedor
print("--- TOTAL VENDIDO POR VENDEDOR ---")
total_por_vendedor = df_ventas.groupby("vendedor")["TotalVenta"].sum()
print(total_por_vendedor)

# 4. Identifique el producto más vendido (en cantidad)
print("\n--- PRODUCTO MÁS VENDIDO (CANTIDAD TOTAL) ---")
producto_mas_vendido = df_ventas.groupby("producto")["cantidad"].sum().idxmax()
print(f"El producto más vendido es: {producto_mas_vendido}")

# 5. Identifique la categoría con mayor ingreso
print("\n--- CATEGORÍA CON MAYOR INGRESO ---")
categoria_mayor_ingreso = df_ventas.groupby("categoría")["TotalVenta"].sum().idxmax()
print(f"La categoría con más ingresos es: {categoria_mayor_ingreso}")

# 6. Ordene las ventas de mayor a menor (según TotalVenta)
print("\n--- VENTAS ORDENADAS DE MAYOR A MENOR ---")
ventas_ordenadas = df_ventas.sort_values(by="TotalVenta", ascending=False)
print(ventas_ordenadas)
```

```python
#Ejercicio 18: Limpieza de datos con pandas
#• Cree un DataFrame con datos incompletos de clientes.
#• Incluya nombre, edad, correo, ciudad y teléfono.
#• Identifique valores nulos.
#• Cuente los valores nulos por columna.
#• Reemplace ciudades vacías con "No especificada".
#• Elimine registros que no tengan correo.
#• Muestre el DataFrame limpio.

import pandas as pd


datos_clientes = {
    "nombre": ["Juan", "Marta", "Pedro", "Sofía", "Lucas"],
    "edad": [25, None, 34, 22, None],
    "correo": ["juan@mail.com", "marta@mail.com", None, "sofia@mail.com", "lucas@mail.com"],
    "ciudad": ["Tegucigalpa", None, "San Pedro", "La Ceiba", None],
    "teléfono": [998877, 887766, None, 332211, 776655]
}

df_clientes = pd.DataFrame(datos_clientes)
print("--- DATOS ORIGINALES CON NULOS ---")
print(df_clientes)


print("\n--- MAPA DE VALORES NULOS (TRUE/FALSE) ---")
print(df_clientes.isnull())


print("\n--- CANTIDAD DE NULOS POR COLUMNA ---")
print(df_clientes.isnull().sum())



df_clientes["ciudad"] = df_clientes["ciudad"].fillna("No especificada")

df_clientes = df_clientes.dropna(subset=["correo"])

print("\n--- DATOS LIMPIOS FINALES ---")
print(df_clientes)
```

```python
#Ejercicio 19: Análisis de préstamos
#• Cree un DataFrame con solicitudes de préstamo.
#• Incluya cliente, monto solicitado, plazo, ingreso mensual y estado de solicitud.
#• Calcule el monto promedio solicitado.
#• Cuente solicitudes por estado.
#• Muestre únicamente préstamos aprobados.
#• Muestre clientes con monto solicitado mayor a 50,000.
#• Agrupe por estado y calcule el promedio del monto solicitado.


prestamos = [
    {"cliente": "Andrés", "monto": 20000, "plazo": 12, "ingreso": 1500, "estado": "Aprobado"},
    {"cliente": "Beatriz", "monto": 65000, "plazo": 36, "ingreso": 4500, "estado": "Aprobado"},
    {"cliente": "César", "monto": 15000, "plazo": 24, "ingreso": 1200, "estado": "Rechazado"},
    {"cliente": "Diana", "monto": 80000, "plazo": 48, "ingreso": 5000, "estado": "Pendiente"},
    {"cliente": "Eduardo", "monto": 45000, "plazo": 24, "ingreso": 3000, "estado": "Aprobado"},
    {"cliente": "Fabiola", "monto": 120000, "plazo": 60, "ingreso": 7500, "estado": "Rechazado"}
]


suma_montos = 0
for p in prestamos:
    suma_montos += p["monto"]
print(f"Monto promedio solicitado: ${suma_montos / len(prestamos):.2f}\n")


conteos_estado = {}
for p in prestamos:
    est = p["estado"]
    if est not in conteos_estado:
        conteos_estado[est] = 0
    conteos_estado[est] += 1
print("--- CANTIDAD DE SOLICITUDES POR ESTADO ---")
print(conteos_estado)


print("\n--- PRÉSTAMOS APROBADOS ---")
for p in prestamos:
    if p["estado"] == "Aprobado":
        print(f"Cliente: {p['cliente']} | Monto: ${p['monto']}")


print("\n--- PRÉSTAMOS MAYORES A $50,000 ---")
for p in prestamos:
    if p["monto"] > 50000:
        print(f"Cliente: {p['cliente']} | Monto: ${p['monto']}")

agrupado = {}
for p in prestamos:
    est = p["estado"]
    if est not in agrupado:
        agrupado[est] = {"suma": 0, "cantidad": 0}
    agrupado[est]["suma"] += p["monto"]
    agrupado[est]["cantidad"] += 1

print("\n--- MONTO PROMEDIO POR ESTADO ---")
for estado, datos in agrupado.items():
    promedio = datos["suma"] / datos["cantidad"]
    print(f"Estado: {estado} | Promedio: ${promedio:.2f}")
```

```python
#Ejercicio 20: Mini análisis de datos tipo empresa
#• Cree un DataFrame con información de empleados.
#• Incluya nombre, departamento, cargo, salario, años de experiencia y ciudad.
#• Muestre los primeros registros.
#• Muestre estadísticas generales de salarios.
#• Calcule el salario promedio por departamento.
#• Encuentre el empleado con salario más alto.
#• Muestre empleados con más de 5 años de experiencia.
#• Ordene empleados por salario.
#• Cree una columna NivelExperiencia: Junior, Intermedio o Senior.


empleados = [
    {"nombre": "Roberto", "depto": "IT", "cargo": "Desarrollador", "salario": 2500, "experiencia": 6, "ciudad": "Tegucigalpa"},
    {"nombre": "Lucía", "depto": "HR", "cargo": "Gerente", "salario": 3000, "experiencia": 8, "ciudad": "San Pedro"},
    {"nombre": "Alejandro", "depto": "IT", "cargo": "Soporte", "salario": 1500, "experiencia": 2, "ciudad": "Choluteca"},
    {"nombre": "Gabriela", "depto": "Ventas", "cargo": "Ejecutivo", "salario": 1800, "experiencia": 4, "ciudad": "Tegucigalpa"},
    {"nombre": "Ricardo", "depto": "Ventas", "cargo": "Gerente", "salario": 3200, "experiencia": 10, "ciudad": "San Pedro"},
    {"nombre": "Valeria", "depto": "IT", "cargo": "Desarrollador", "salario": 2700, "experiencia": 3, "ciudad": "Tegucigalpa"}
]


print("--- PRIMEROS REGISTROS ---")
for i in range(3):
    print(empleados[i])


salarios = [emp["salario"] for emp in empleados]
print(f"\nSalario Máximo: ${max(salarios)} | Salario Mínimo: ${min(salarios)}")


depto_salarios = {}
for emp in empleados:
    d = emp["depto"]
    if d not in depto_salarios:
        depto_salarios[d] = {"suma": 0, "cantidad": 0}
    depto_salarios[d]["suma"] += emp["salario"]
    depto_salarios[d]["cantidad"] += 1

print("\n--- SALARIO PROMEDIO POR DEPTO ---")
for depto, datos in depto_salarios.items():
    print(f"Departamento: {depto} | Promedio: ${datos['suma']/datos['cantidad']:.2f}")


empleado_rico = max(empleados, key=lambda x: x["salario"])
print(f"\nEmpleado con salario más alto: {empleado_rico['nombre']} (${empleado_rico['salario']})")


print("\n--- MÁS DE 5 AÑOS DE EXPERIENCIA ---")
for emp in empleados:
    if emp["experiencia"] > 5:
        print(f"{emp['nombre']} tiene {emp['experiencia']} años de experiencia.")


empleados.sort(key=lambda x: x["salario"])


for emp in empleados:
    exp = emp["experiencia"]
    if exp < 3:
        emp["NivelExperiencia"] = "Junior"
    elif exp <= 5:
        emp["NivelExperiencia"] = "Intermedio"
    else:
        emp["NivelExperiencia"] = "Senior"

print("\n--- LISTA ORDENADA Y CLASIFICADA FINAL ---")
for emp in empleados:
    print(f"Nombre: {emp['nombre']} | Salario: ${emp['salario']} | Nivel: {emp['NivelExperiencia']}")
```
