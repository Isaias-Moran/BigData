# Análisis de un Caso Real de Big Data: Infraestructura Vial

## Caso de estudio: Uso de datos de movilidad y sensores para la construcción de puentes o calles alternativa basada en datos diarios, con la finalidad de mejorer el trafico.



### 1. ¿Qué datos genera?
* **¿Qué son?:** Ubicaciones de GPS de los celulares de los conductores, videos de las cámaras de la calle.
* **Cantidad y velocidad:** Son millones de datos que entran sin parar las 24 horas del día, registrando el paso de cada carro, moto o camión.
* **Idea clave:** No solo miden el tráfico cuando la calle está llena, sino que registran el movimiento real a cualquier hora del día y de la noche.



### 2. ¿Quién los usa?
* **Alcaldías y Ministerios de Obras Públicas:** Para decidir si vale la pena gastar el presupuesto en un puente, ampleacion de carriles o calles completamente nuevaspor esas rutas.
* **Ingenieros y diseñadores:** Para planificar el tamaño del puente y los carriles que se necesitan.



### 3. ¿Qué problema resuelve?
* **Evita adivinar:** En lugar de asumir que se ocupa un puente solo porque la calle se llena a las 6:00 p.m., el sistema demuestra si el flujo de carros sigue siendo insostenible a otras horas o los fines de semana. 
* **Ahorra dinero:** Si los datos muestran que el tráfico pesado (como camiones de carga) pasa de madrugada, el puente se diseña con la resistencia exacta, evitando que la obra colapse antes de tiempo o que se gaste de más.



### 4. ¿Por qué no bastaría Excel?
* **Por el tamaño:** Excel se traba o se cierra si intentas meterle millones de ubicaciones de GPS y videos guardados. No le cabe tanta información.
* **Por la variedad:** Excel solo entiende tablas con filas y columnas; no puede procesar el video de una cámara para contar cuántos carros pasaron por sí solo, ni cruzar las rutas de los mapas en tiempo real.


### Justificación de por qué se considera Big Data
* **Volumen masivo:** No estamos analizando una muestra de 100 carros. El sistema procesa los registros de ubicación de miles de conductores y flujos de video continuos todos los días.
* **Velocidad en tiempo real:** Los datos no se recolectan una vez al año; entran en un flujo constante las 24 horas del día, los 7 días de la semana, directo de los satélites y camaras.
* **Variedad de formatos:** Combina datos que no tienen nada que ver entre sí: coordenadas de mapas (GPS) y archivos de video de las cámaras
